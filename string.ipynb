```python
while True:
    user_input = input("Enter text here: ")
    print(f"Original text is : {user_input}")
    confirmation = input("Are you sure about your input? (yes/no): ")

    if confirmation.lower() == 'yes':
        final_string = user_input
        print(f"Your final message is: {final_string}")

        # left-to-right slice
        sliced_left = "[No left slice performed]" # Initialize with a default message
        try:
            num_from_left_str = input(f"The message is '{final_string}' (length: {len(final_string)}). How many characters do you want to slice from the left? ")
            num_from_left = int(num_from_left_str)

            if num_from_left < 0:
                print("Slice amount cannot be negative. No left slice performed.")
            elif num_from_left > len(final_string):
                print(f"Cannot slice {num_from_left} characters from left. String length is {len(final_string)}. Slicing full string instead.")
                sliced_left = final_string[:]
            else:
                sliced_left = final_string[:num_from_left]
            print(f"Sliced from left (first {num_from_left} characters): '{sliced_left}'")

        except ValueError:
            print("Invalid input for left slice. Please enter an integer. No left slice performed.")
        except Exception as e:
            print(f"An unexpected error occurred during left slicing: {e}. No left slice performed.")

        # right-to-left slice
        sliced_right = "[No right slice performed]"
        try:
            num_from_right_str = input(f"How many characters do you want to slice from the right? ")
            num_from_right = int(num_from_right_str)

            if num_from_right < 0:
                print("Slice amount cannot be negative. No right slice performed.")
            elif num_from_right > len(final_string):
                print(f"Cannot slice {num_from_right} characters from right. String length is {len(final_string)}. Slicing full string instead.")
                sliced_right = final_string[:]
            else:
                sliced_right = final_string[-num_from_right:]
            print(f"Sliced from right (last {num_from_right} characters): '{sliced_right}'")

        except ValueError:
            print("Invalid input for right slice. Please enter an integer. No right slice performed.")
        except Exception as e:
            print(f"An unexpected error occurred during right slicing: {e}. No right slice performed.")

        break
    else:
        print("Please provide the input again.")
```

```python
import getpass

# Predefined correct username and password
correct_username = "admin"
correct_password = "password123"

while True:
    username_input = input("Enter username: ")
    password_input = getpass.getpass("Enter password: ")

    if username_input == correct_username and password_input == correct_password:
        print("Welcome aboard!")
        break
    else:
        print("Username/password is incorrect. Please try again.")
```

```python
max_retries = 3
retries = 0

while retries < max_retries:
    user_input = input("Enter text here: ")
    print(f"Original text is : {user_input}")
    confirmation = input("Are you sure about your input? (yes/no): ")

    if confirmation.lower() == 'yes':
        final_string = user_input + 'thank you'
        print(f"Your final message is: {final_string}")
        break # Exit the loop if confirmed
    else:
        retries += 1
        if retries < max_retries:
            print("Please provide the input again.")
        else:
            print("Maximum retries reached.")
```
