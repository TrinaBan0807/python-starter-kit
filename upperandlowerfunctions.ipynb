```python
user_input = input("Enter text here: ")
print(f"Original text is : {user_input}")
upper_case_text = user_input.upper()
lower_case_text = user_input.lower()
print(f"Uppercase: {upper_case_text}")
print(f"Lowercase: {lower_case_text}")
```
